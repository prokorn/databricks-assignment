```python
# Daily Scheduled Reconciliation Audit
dbutils.widgets.text("catalog", "main", "Catalog Name")
dbutils.widgets.text("schema", "lab_data", "Schema Name")

catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")

bronze_cnt = spark.table(f"{catalog}.{schema}.movies_bronze").count()
silver_cnt = spark.table(f"{catalog}.{schema}.movies_silver").count()
quar_cnt = spark.table(f"{catalog}.{schema}.movies_quarantine").count()

total_processed = silver_cnt + quar_cnt
loss_diff = bronze_cnt - total_processed

print(f"[{catalog}.{schema}] Bronze: {bronze_cnt} | Silver: {silver_cnt} | Quarantine: {quar_cnt}")
print(f"Silent Loss Difference: {loss_diff}")

assert loss_diff == 0, f"DATA LOSS ALERT: {loss_diff} records missing between layers!"
print("STATUS: Reconciliation PASSED")
```
