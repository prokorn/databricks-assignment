```python
# ==========================================
# 04. SCHEMA EVOLUTION DEMONSTRATION
# ==========================================
from pyspark.sql.functions import lit

catalog = "workspace"
schema_name = "default"
silver_scd1_table = f"{catalog}.{schema_name}.wiki_silver_scd1"

# 1. Read existing SCD1 table sample
df_scd1 = spark.table(silver_scd1_table)

# 2. Add a NEW column ('edit_length') to test Schema Evolution
df_new_batch = df_scd1.limit(1) \
    .withColumn("edit_length", lit(450)) \
    .withColumn("comment", lit("updated with new schema field"))

# 3. Append with mergeSchema enabled
df_new_batch.write \
    .format("delta") \
    .mode("append") \
    .option("mergeSchema", "true") \
    .saveAsTable(silver_scd1_table)

print("✅ Step 4 Complete! Schema Evolution successfully applied.")
display(spark.table(silver_scd1_table))
```
