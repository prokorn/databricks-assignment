```python
# ==========================================
# 05. TABLE MAINTENANCE & OPTIMIZATION
# ==========================================

catalog = "workspace"
schema_name = "default"
silver_scd1_table = f"{catalog}.{schema_name}.wiki_silver_scd1"

# 1. Apply Liquid Clustering
spark.sql(f"ALTER TABLE {silver_scd1_table} CLUSTER BY (wiki, id)")

# 2. Compact small files using OPTIMIZE
spark.sql(f"OPTIMIZE {silver_scd1_table}")

# 3. Vacuum obsolete file versions
spark.sql(f"VACUUM {silver_scd1_table}")

print("✅ Step 5 Complete! Liquid Clustering, OPTIMIZE & VACUUM executed successfully.")
```
